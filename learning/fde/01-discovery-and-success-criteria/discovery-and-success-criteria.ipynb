**Warning:** A customer claim does not become measured when copied into a KPI table. A calculation over a modeled input remains modeled.

**Common Pitfalls**

| | Pattern | Why it matters |
|---|---|---|
| Wrong | Start with model, agent, database, and demo choices | The workshop can look productive while covering 0 of 11 decision dimensions |
| Right | Start with users, workflow, baseline, failure cost, authority, and evidence owners | Architecture choices inherit a reviewable reason and a decision owner |

**Quick Health Check:** The next cell must expose the missing dimensions, resolve the prediction by name, and let you change exactly one discovery dimension without hiding the remaining gaps.

**Reflection:** Adding one workshop question improves coverage by one, not by magic. The first durable repair is to map stakeholders and decision rights so each later artifact has an authorized owner.

````python
# ── Measure the solution-first coverage failure ──────────────────────────
solution_dimensions = {"solution", "demo"}
required_dimensions = {"users", "workflow", "baseline", "outcomes", "failure_cost", "authority", "data", "identity", "constraints", "acceptance", "ownership"}
missing = sorted(required_dimensions - solution_dimensions)
print(f"Discovery coverage: {len(required_dimensions & solution_dimensions)}/{len(required_dimensions)}")
print(f"Missing: {', '.join(missing)}")
print("Prediction confirmed: missing problem and authority evidence invalidates the workshop.")

new_dimension = "workflow"  # CHANGE THIS: choose one missing dimension
before = len(required_dimensions & solution_dimensions)
after = len(required_dimensions & (solution_dimensions | {new_dimension}))
assert new_dimension in required_dimensions and after == before + 1
print(f"Your turn PASS: coverage moves from {before} to {after}; the remaining gaps stay visible.")
````
