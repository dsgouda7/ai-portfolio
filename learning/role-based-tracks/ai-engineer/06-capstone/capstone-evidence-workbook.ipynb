# Riverside Release Review: AI Engineer Capstone

> **The story:** AI application releases bind more than model weights. Data, adapter, tokenizer, prompt, retrieval index, evaluators, runtime, gateway behavior, operations, and feedback can all change the result. Your task is to make one Riverside release decision another engineer can audit.
>
> **Where you are:** The AI Engineer route produced separate operational evidence surfaces. This workbook joins them through contracts and links. It copies no upstream analysis code and assumes no shared kernel state.
>
> **Notation:** $r_c$ is the candidate; $r_b$ is the baseline; $R(r_c)$ is the rollback target; $G_i$ is gate $i$; $E(x)$ is the evidence class; $D \in \{promote, hold, reject\}$.

> **Workbook state:** Markdown-only and structurally validated. Fill a candidate copy with artifact links and review decisions; there is no executable analysis code in this workbook.

## 0 · The Challenge

> **The mission:** Riverside House must decide whether `rel-riv-002` may advance one release stage without allowing an aggregate improvement, a source-only artifact, or an unvalidated cloud claim to compensate for a failed boundary.

**What we know so far:**

- Chapters 01-05 define separate contracts for data, prompt, operations, release lineage, and feedback.
- The RAG and evaluation chapters separate retrieval failure from generation and policy failure.
- **But separate green-looking artifacts do not yet prove one coherent release is safe to advance.**

**What's blocking us:** The package can mix release IDs, omit rollback evidence, label modeled architecture as `LIVE_AZURE`, or hide a critical-slice failure behind an aggregate score. Any one of those defects makes the final decision unauditable.

**What this workbook unlocks:** A staged, non-compensating review in which every gate points to retained evidence, every unsupported claim remains visible, and the final wording is no stronger than the evidence class allows.

## C7 - Final Decision and Defense

The final decision is a conjunction across required gates:

$$D=promote \iff \bigwedge_i G_i=pass$$

In plain language: promotion is allowed only when every gate required for the requested stage passes. `hold` is correct when evidence is incomplete or underpowered; `reject` is correct when a boundary fails.

### Common Pitfalls

| | Pattern | Why it matters |
|---|---|---|
| Wrong | Promote because most gates are green | Required gates are non-compensating |
| Wrong | Treat a placeholder or source file as measured proof | Evidence class and conclusion no longer match |
| Wrong | Collapse retrieval and generation into one score | The remediation owner becomes guesswork |
| Right | Link every verdict to a release-scoped artifact and preserve unsupported claims | Another engineer can reproduce the decision boundary |

### Quick Health Check

Before defense, verify one candidate ID across every artifact; one accepted rollback target; separate retrieval, generation, safety, operations, and cloud gates; no placeholders labeled as measured; and a final decision consistent with every non-compensating gate.

### Key Takeaways

- Evidence strength determines release stage.
- A failed boundary cannot be averaged away.
- Missing evidence supports `hold`, not optimistic promotion.
- The narrowest true claim is the strongest defensible claim.

### Self-assessment

Apply the [assessment rubric](assessment-rubric.md). Record raw score, score caps, final score, and every deduction. Do not award points for a green result; award points for trustworthy evidence.
