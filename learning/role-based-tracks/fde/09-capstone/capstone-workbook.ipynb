````python
# ── Flag Unsupported Claim Language ─────────────────────────────────────────
unsupported_terms = {
    "production-ready", "secure", "compliant", "resident",
    "highly available", "scalable", "within sla", "cost optimized",
}
candidate_claims = [
    {
        "claim_id": "CLM-CAP-PROBE-001",
        "statement": "Riverside is region-compliant.",
        "class": None,
        "evidence_ref": None,
        "limitations": None,
        "revalidate_on": None,
    },
    {
        "claim_id": "CLM-CAP-PROBE-002",
        "statement": (
            "The source design carries region metadata; service-by-service "
            "residency and deployed configuration remain externally unvalidated."
        ),
        "class": "External validation required",
        "evidence_ref": "MAP-GAP-003 and SEC-CAP-03",
        "limitations": "No deployed service or data-flow evidence.",
        "revalidate_on": "Approved service/region review and live configuration evidence",
    },
]

def claim_findings(claim: dict) -> list[str]:
    statement = claim["statement"].lower()
    findings = [term for term in unsupported_terms if term in statement]
    for field in ("class", "evidence_ref", "limitations", "revalidate_on"):
        if not claim.get(field):
            findings.append(f"missing:{field}")
    return sorted(findings)

for claim in candidate_claims:
    findings = claim_findings(claim)
    status = "BLOCK" if findings else "REVIEW"
    print(f"{status}: {claim['claim_id']} -> {findings or ['bounded claim record present']}")

print("Prediction check: the first statement is blocked; the second still requires named review.")
````

## 5 · Identity, Threat, Residency, and Controls

> A contractor left Riverside yesterday, but their cached entitlement still says `active`. They request policy for a title they were never assigned. How does the system deny access before retrieval?

Authentication only says who presented a credential. The deny path must also check current tenant, role, employment state, purpose, title assignment, entitlement freshness, region, requested tool, and deletion state using trusted server-derived context.

### Deny scenario checklist

- [ ] The gateway ignores client-supplied tenant, role, ACL, and title authority.
- [ ] Current identity and entitlement state is read from the approved source or fails closed when freshness is unknown.
- [ ] Tenant, purpose, role, title assignment, region, and deletion state are checked before retrieval.
- [ ] Server-built filters apply before ranking, and returned records are checked again.
- [ ] Tool authorization evaluates the concrete PageTurn payload independently of model intent.
- [ ] The deny event preserves approved IDs, versions, reason, and trace context without copying customer content.
- [ ] Re-enablement requires the named authority and repeated negative tests.

````python
# ── Enforce Non-Compensating Release Gates ──────────────────────────────────
release_gates = [
    {"gate_id": "EVAL-CAP-DATA", "status": None, "critical": False},
    {"gate_id": "EVAL-CAP-RETRIEVAL", "status": None, "critical": False},
    {"gate_id": "EVAL-CAP-GENERATION", "status": None, "critical": False},
    {"gate_id": "EVAL-CAP-AUTHORIZATION", "status": None, "critical": True},
    {"gate_id": "EVAL-CAP-DELETION", "status": None, "critical": True},
    {"gate_id": "EVAL-CAP-DUPLICATE-COMMIT", "status": None, "critical": True},
    {"gate_id": "EVAL-CAP-OPERATIONS", "status": None, "critical": False},
]

def promotion_decision(gates: list[dict]) -> str:
    if any(gate["critical"] and gate["status"] == "FAIL" for gate in gates):
        return "REJECT"
    if any(gate["status"] in (None, "HOLD") for gate in gates):
        return "HOLD"
    if any(gate["status"] == "FAIL" for gate in gates):
        return "REJECT"
    return "PROMOTE_TO_NAMED_NEXT_STAGE_ONLY"

authored_gate_decision = promotion_decision(release_gates)
print(f"Authored gate decision: {authored_gate_decision}")
print(f"Prediction closed: {authored_gate_decision} - unset evidence cannot authorize exposure.")
````
